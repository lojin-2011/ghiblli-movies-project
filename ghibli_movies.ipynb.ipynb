```python
#importing libraries
import pandas as pd
import requests
import matplotlib.pyplot as plt
```

```python
#srep 2 - retrieve data from the Github API
url='https://ghibliapi.vercel.app/films'
response=requests.get(url)
data=response.json()
```

```python
#extract columns
df1=movies[['title','director','producer','release_date','running_time','rt_score']]
df1.head()
```

```python
#clean data step 1:nulls
print(df1.isnull().sum)
```

```python
df1 = df1.drop_duplicates()
```

```python
#change dtypes
df1['release_date']=pd.to_datetime(df1['release_date'],format='%Y')
```

```python
df1["rt_score"]=pd.to_numeric(df1["rt_score"])
```

```python
df1.rename(
    columns={
        "release_date":"release_year",
        "running_time":"duration_minutes",
        "rt_score":"rating"
    },
    inplace=True
)
df1.head()
```

```python
%load_ext sql

%config SqlMagic.style = "PLAIN_COLUMNS"

%sql sqlite:///ghibli_movies.db
```

```python
import sqlite3

conn = sqlite3.connect("ghibli_movies.db")

movies.to_sql(
    "MOVIES",
    conn,
    if_exists="replace",
    index=False
)
```

```python
%%sql
SELECT*FROM MOVIES
```

```python
%%sql
SELECT *
FROM MOVIES
WHERE rating > 90;
```

```python
%%sql
SELECT title, director, rating
FROM MOVIES
WHERE rating > 90
AND duration_minutes > 100;
```

```python
%%sql
SELECT title, director, rating
FROM MOVIES
WHERE rating > 90
OR duration_minutes > 120;
```

```python
%%sql
SELECT title, director, rating
FROM MOVIES
WHERE NOT director = 'Hayao Miyazaki';
```

```python
%%sql
SELECT title, director, rating
FROM MOVIES
ORDER BY rating DESC
LIMIT 5;
```

```python
%%sql
SELECT
    COUNT(*) AS total_movies,
    AVG(rating) AS average_rating,
    AVG(duration_minutes) AS average_duration
FROM MOVIES;
```

```python
%%sql
SELECT release_year, COUNT(*) AS movie_count
FROM MOVIES
GROUP BY release_year
ORDER BY release_year;
```

```python
%%sql
SELECT director, COUNT(*) AS movie_count
FROM MOVIES
GROUP BY director
HAVING COUNT(*) > 1
ORDER BY movie_count DESC;
```

```python
df1['release_year'] = pd.to_numeric(df1['release_year'])
movies_per_year = df1.groupby('release_year').size()
plt.figure(figsize=(10,5))
plt.bar(movies_per_year.index, movies_per_year.values)
plt.xlabel("Release Year")
plt.ylabel("Number of Movies")
plt.title("Number of Ghibli Movies by Release Year")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.bar(movies['title'], movies['rating'])
plt.xlabel("Movie")
plt.ylabel("Rating")
plt.title("Ghibli Movies Ratings")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(movies['duration_minutes'], movies['rating'])
plt.xlabel("Duration (Minutes)")
plt.ylabel("Rating")
plt.title("Movie Duration vs Rating")
plt.show()
```

The analysis shows that Ghibli movies have different ratings and durations.
The SQL queries helped identify highly rated movies.
The release year analysis shows the distribution of movies over time.
The visualizations made it easier to compare movies by rating, duration, and release year.
